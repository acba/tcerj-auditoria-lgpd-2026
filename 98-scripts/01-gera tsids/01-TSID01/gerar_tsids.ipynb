```python
import pandas as pd
import os.path

from docxtpl import DocxTemplate
from docx2pdf import convert

if not os.path.exists('docx/'):
    os.makedirs('docx/')

if not os.path.exists('pdf/'):
    os.makedirs('pdf/')
    
base = DocxTemplate("TSID01 - Base.docx")
consolidado = pd.read_excel('consolidado.xlsx')

diatsid = '20'

for index, row in consolidado.iterrows():
    print(f"Processando {row['sigla']}")
    
    filename = f"TSID01 - {row['sigla'].strip()}".replace('/', '-')
    
    if os.path.isfile(f'pdf/{filename}.pdf'):
        print('Já processado')
        
    else:
        context = { 
            'nof' : ('0000' + str(row['num_oa']))[-4:],
            'diatsid': diatsid,
        }
        base.render(context)


        base.save(f'docx/{filename}.docx')
        convert(f'docx/{filename}.docx', f'pdf/{filename}.pdf')
```

```python
import os
import re
import csv

# Diretório contendo os PDFs
DIRETORIO = r"F:\CAD-TI\AUDITORIAS\2026.2-Auditoria LGPD\99-Gestão\01-Ofícios de Apresentação\assinados"
DIRETORIO_SAIDA = r"F:\CAD-TI\AUDITORIAS\2026.2-Auditoria LGPD\98-scripts\01-gera tsids\01-TSID01"

# Regex para extração
PADRAO = re.compile(
    r'_PREFEITURA (.*?)_(\d{4})_2026_assinado',
    re.IGNORECASE
)

registros = []

for arquivo in os.listdir(DIRETORIO):
    caminho = os.path.join(DIRETORIO, arquivo)

    if not os.path.isfile(caminho):
        continue

    m = PADRAO.search(arquivo)

    if m:
        orgao = m.group(1).strip()
        nof = m.group(2)

        registros.append({
            "orgao": orgao,
            "nof": nof
        })

# Grava CSV
saida_csv = os.path.join(DIRETORIO_SAIDA, "resultado.csv")

with open(saida_csv, "w", newline="", encoding="utf-8-sig") as f:
    writer = csv.DictWriter(
        f,
        fieldnames=["orgao", "nof"],
        delimiter=";"
    )
    writer.writeheader()
    writer.writerows(registros)

print(f"{len(registros)} registros gravados em {saida_csv}")
```

```python
import os
import re
import unicodedata
from difflib import get_close_matches

DIRETORIO = r"F:\CAD-TI\AUDITORIAS\2026.2-Auditoria LGPD\99-Gestão\01-Ofícios de Apresentação\assinados"

MUNICIPIOS = [
    "ANGRA DOS REIS", "APERIBÉ", "ARARUAMA", "AREAL",
    "ARMAÇÃO DOS BÚZIOS", "ARRAIAL DO CABO", "BARRA DO PIRAÍ",
    "BARRA MANSA", "BELFORD ROXO", "BOM JARDIM",
    "BOM JESUS DO ITABAPOANA", "CABO FRIO",
    "CACHOEIRAS DE MACACU", "CAMBUCI",
    "CAMPOS DOS GOYTACAZES", "CANTAGALO", "CARAPEBUS",
    "CARDOSO MOREIRA", "CARMO", "CASIMIRO DE ABREU",
    "COMENDADOR LEVY GASPARIAN", "CONCEIÇÃO DE MACABU",
    "CORDEIRO", "DUAS BARRAS", "DUQUE DE CAXIAS",
    "ENGENHEIRO PAULO DE FRONTIN", "GUAPIMIRIM",
    "IGUABA GRANDE", "ITABORAÍ", "ITAGUAÍ", "ITALVA",
    "ITAOCARA", "ITAPERUNA", "ITATIAIA", "JAPERI",
    "LAJE DO MURIAÉ", "MACAÉ", "MACUCO", "MAGÉ",
    "MANGARATIBA", "MARICÁ", "MENDES", "MESQUITA",
    "MIGUEL PEREIRA", "MIRACEMA", "NATIVIDADE",
    "NILÓPOLIS", "NITERÓI", "NOVA FRIBURGO",
    "NOVA IGUAÇU", "PARACAMBI", "PARAÍBA DO SUL",
    "PARATY", "PATY DO ALFERES", "PETRÓPOLIS",
    "PINHEIRAL", "PIRAÍ", "PORCIÚNCULA",
    "PORTO REAL", "QUATIS", "QUEIMADOS", "QUISSAMÃ",
    "RESENDE", "RIO BONITO", "RIO CLARO",
    "RIO DAS FLORES", "RIO DAS OSTRAS",
    "SANTA MARIA MADALENA", "SANTO ANTÔNIO DE PÁDUA",
    "SAPUCAIA", "SAQUAREMA", "SÃO GONÇALO",
    "SEROPÉDICA", "SILVA JARDIM", "SUMIDOURO",
    "SÃO FIDÉLIS", "SÃO FRANCISCO DE ITABAPOANA",
    "SÃO JOSÉ DO VALE DO RIO PRETO",
    "SÃO JOÃO DA BARRA", "SÃO JOÃO DE MERITI",
    "SÃO PEDRO DA ALDEIA", "SÃO SEBASTIÃO DO ALTO",
    "SÃO JOSÉ DE UBÁ", "TANGUÁ",
    "TERESÓPOLIS", "TRAJANO DE MORAES",
    "TRÊS RIOS", "VALENÇA", "VARRE-SAI",
    "VASSOURAS", "VOLTA REDONDA"
]


def normalizar(texto):
    texto = unicodedata.normalize('NFKD', texto)
    texto = ''.join(c for c in texto if not unicodedata.combining(c))
    texto = texto.upper()

    # mantém apenas letras, números, espaço e hífen
    texto = re.sub(r'[^A-Z0-9 \-]', '', texto)

    # remove espaços duplicados
    texto = re.sub(r'\s+', ' ', texto).strip()

    return texto


# índice para busca aproximada
municipios_norm = {
    normalizar(m): m
    for m in MUNICIPIOS
}

for nome_arquivo in os.listdir(DIRETORIO):

    caminho_antigo = os.path.join(DIRETORIO, nome_arquivo)

    if not os.path.isfile(caminho_antigo):
        continue

    # Extrai município
    m = re.search(
        r'PREFEITURA\s+(.*?)_?\d{4}_2026_assinado',
        nome_arquivo,
        re.IGNORECASE
    )

    if not m:
        continue

    municipio_extraido = m.group(1).strip()
    municipio_norm = normalizar(municipio_extraido)

    correspondencia = get_close_matches(
        municipio_norm,
        municipios_norm.keys(),
        n=1,
        cutoff=0.6
    )

    if not correspondencia:
        print(f"Não identificado: {nome_arquivo}")
        continue

    municipio_correto = municipios_norm[correspondencia[0]]

    novo_nome = nome_arquivo.replace(
        municipio_extraido,
        municipio_correto
    )

    if novo_nome != nome_arquivo:
        caminho_novo = os.path.join(DIRETORIO, novo_nome)

        print(f"{nome_arquivo}")
        print(f" -> {novo_nome}")

        os.rename(caminho_antigo, caminho_novo)

print("Concluído.")
```
